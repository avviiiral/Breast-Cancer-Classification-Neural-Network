```python
import pandas as pd
import numpy as np  
import matplotlib.pyplot as plt
import sklearn.datasets 
from sklearn.model_selection import train_test_split
```

```python
#loading dataset
breast_cancer_dateset = sklearn.datasets.load_breast_cancer()
```

```python
print(breast_cancer_dateset)
```

```python
# load the data to dataframe
data_frame = pd.DataFrame(breast_cancer_dateset.data, columns=breast_cancer_dateset.feature_names)
```

```python
# print first 5  rows
data_frame.head()
```

```python
# adding the 'target' column to the data frame
data_frame['label'] = breast_cancer_dateset.target
```

```python
# print last 5 rows of the dataframe
data_frame.tail()
```

```python
# number of rows and columns
data_frame.shape

```

```python
# getting some information about the data
data_frame.info()   
```

```python
# checking for missing values
data_frame.isnull().sum()
```

```python
# statistical measures about the data
data_frame.describe()
```

```python
# checking the distribution of target variable
data_frame['label'].value_counts()
```

```python
data_frame.groupby('label').mean()
```

```python
# Seprating featurs and target
X = data_frame.drop(columns='label', axis=1)
Y = data_frame['label']
```

```python
print(X)
```

```python
print(Y)
```

```python
# Setting data into training and test dataset
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
# Standardizing the data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_std = scaler.fit_transform(X_train)
X_test_std = scaler.transform(X_test)
```

```python
# importing tensorflow and keras
import tensorflow as tf
from tensorflow import keras
tf.random.set_seed(3)
```

```python
# setting layer of neural network
model = keras.Sequential([
    keras.layers.Flatten(input_shape = (30,)),
    keras.layers.Dense(20, activation='relu'),
    keras.layers.Dense(2, activation='sigmoid')
])
```

```python
# compiling the neural network
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics = ['accuracy'])
```

```python
# training the neural network
history = model.fit(X_train_std, Y_train, validation_split=0.1, epochs=10)
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title('Model Accuracy')
plt.ylabel('accuracy')
plt.xlabel('epoch')

plt.legend(['training data','validation data'], loc = 'lower right')
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title('model loss')
plt.ylabel('loss')
plt.xlabel('epoch')

plt.legend(['training data', 'validation data'], loc = 'upper right')
```

```python
loss, accuracy = model.evaluate(X_test_std, Y_test)
print(accuracy)
```
